### LIMPIEZA

```python
# Comprobamos que no haya valores duplicados:
print(f'Valores duplicados: {df.duplicated().sum()}.')
```

```python
# Tenemos valores nulos:
    # Sacamos el df con los valores nulos
df_nulos = df.isnull().sum().reset_index()
    # Cambiamos el nombre de las columnas
df_nulos.columns= ['Columnas', 'Valores_Nulos(%)']
    # Nos quedamos con las columnas que tengan valores distintos de 0
df_nulos = df_nulos[df_nulos['Valores_Nulos(%)'] != 0]
    # Añadimos una columna con los porcentajes resultantes
porcentaje_nulos = round(df_nulos['Valores_Nulos(%)']/df.shape[0]*100,2)
df_nulos['Porcentaje_Nulos'] = porcentaje_nulos
df_nulos
```

```python
# IMPUTACIÓN DE VALORES
# Son variables CATEGÓRICAS con menos del 5% de nulos:
# volvemos a ver los valores únicos:
columnas_cat_nulos = ['Teacher_Quality','Parental_Education_Level','Distance_from_Home']
for col in df[columnas_cat_nulos].columns:
    print(f"Las categorías que tenemos para la columna {col} son:")
    display(df[col].value_counts().reset_index())
    print('_' * 100)
```

```python
display(df[columnas_cat_nulos].describe().T)
print('Claramente se ve que la MODA en todas las categorías es clara.')
```

```python
# Rellenaremos los datos faltantes por la MODA de cada categoría
# iteramos por la lista creada en el paso anterior:
for columna in columnas_cat_nulos:
    # calculamos la moda
    moda = df[columna].mode()[0]    
    # método fillna para reemplazar los valores nulos por la moda
    df[columna] = df[columna].fillna(moda)

# chequeamos si se han eliminado los nulos
print("Después del reemplazo usando 'fillna' quedan los siguientes nulos")
df[columnas_cat_nulos].isnull().sum()
```

### Insertamos columna de niveles de rendimiento:  

5 NIVELES  
- Exam_Score

```python
round(df['Exam_Score'].describe(),2)
# mínimo --> 55
# máximo --> 101
```

Nivel más bajo de rendimiento:
1) Nivel 1 --> bajo rendimiento  -------------------->  55 - 64  
Se establecen tres niveles intermedios:  
2) Nivel 2 --> intermedio bajo  -------------------->  65 -73 
3) Nivel 3 --> intermedio central  -------------------->  74 - 82  
4) Nivel 4 --> intermedio alto   -------------------->  83 - 91  
Finalmente, el nivel más alto:  
5) Nivel 5: alto rendimiento  -------------------->  92 - 101 

```python
# sacamos la columna 'Exam_Score'
exam_score = list(df['Exam_Score']) # array --> lista

# bucle vamos obteniendo los grupos de categorías
perf_levels= []

for v in exam_score:
    if 55 <= v <= 64:
        perf_levels.append(1)
    if 65 <= v <= 73:
        perf_levels.append(2)
    if 74 <= v <= 82:
        perf_levels.append(3)
    if 83 <= v <= 91:
        perf_levels.append(4)
    if 92 <= v <= 101:
        perf_levels.append(5)

# comprobamos
print(f'Total de elementos del df: {df.shape[0]}.')
print(f'Total de elementos de la lista: {len(perf_levels)}.')
```

```python
# añadimos la columna al df:
df['Performance_Levels'] = perf_levels
df.head(2)
```

- Previus_Score

```python
round(df['Previous_Scores'].describe(),2)
# mínimo --> 50
# máximo --> 100
```

Nivel más bajo de rendimiento:
1) Nivel 1 --> bajo rendimiento  -------------------->  50 - 60  
Se establecen tres niveles intermedios:  
2) Nivel 2 --> intermedio bajo  -------------------->  61 -70 
3) Nivel 3 --> intermedio central  -------------------->  71 - 79  
4) Nivel 4 --> intermedio alto   -------------------->  80 - 89  
Finalmente, el nivel más alto:  
5) Nivel 5: alto rendimiento  -------------------->  90 - 100

```python
# sacamos la columna 'Exam_Score'
prev_score = list(df['Previous_Scores']) # array --> lista

# bucle vamos obteniendo los grupos de categorías
perf_prev_levels= []

for v in prev_score:
    if 50 <= v <= 60:
        perf_prev_levels.append(1)
    if 61 <= v <= 70:
        perf_prev_levels.append(2)
    if 71 <= v <= 79:
        perf_prev_levels.append(3)
    if 80 <= v <= 89:
        perf_prev_levels.append(4)
    if 90 <= v <= 100:
        perf_prev_levels.append(5)

# comprobamos
print(f'Total de elementos del df: {df.shape[0]}.')
print(f'Total de elementos de la lista: {len(perf_prev_levels)}.')
```

```python
# añadimos la columna al df:
df['Previous_Perf_Levels'] = perf_prev_levels
df.head(2)
```

```python
# Valores únicos --> columnas rendimienro
columnas_cat_nulos = ['Previous_Perf_Levels','Performance_Levels']
for col in df[columnas_cat_nulos].columns:
    print(f"Las categorías que tenemos para la columna {col} son:")
    display(df[col].value_counts().reset_index())
    print('_' * 100)
```
